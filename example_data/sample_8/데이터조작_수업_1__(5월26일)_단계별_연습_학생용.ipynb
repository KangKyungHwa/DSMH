#### -컬럼명과 인덱스로 데이터 추출: df['column'], df[['column']], filter, loc, iloc

```python
print(df[ '              '], '\n')              #지점명 별로 데이터 추출하기
print(df[['일시', '지점명']], '\n')
print(df.filter(['일시', '지점명']))
```

#### -loc:pandas DataFrame에서 “행과 열을 라벨(label) 기준으로 선택하는 인덱서"

```python
print(df.loc[0, :], '\n')    #0행의 전체 열에 대한 데이터 추출
print(df.loc[           , ' '], '\n') #0행의 '일시' 데이터 추출
print(df.loc[0:6, ['일시', '지점명', '평균기온(°C)']])
```

```python
print(df.iloc[0, 2], '\n')   #0행의 인덱스 2인 열 추출('일시')
print(df.iloc[0:6, 2], '\n') #0~5행의 인덱스 2인 열('일시') 데이터 추출
print(df.iloc[0:6, [2, 1, 3]])# '일시' --> 2, '지점명'--> 1, '평균기온'--> 3
```
